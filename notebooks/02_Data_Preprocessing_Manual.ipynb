```python
# ==========================================
# STEP 1: Import Required Libraries
# ==========================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder

print("✅ All libraries imported successfully!")
```

```python
# ==========================================
# STEP 2: Load Dataset
# ==========================================

df = pd.read_excel("../Data/fraud_dataset.xlsx")

print("✅ Dataset Loaded Successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# ==========================================
# STEP 3: Dataset Information
# ==========================================

print(df.info())
```

```python
# ==========================================
# STEP 4: Check Missing Values
# ==========================================

print("Missing Values")
print(df.isnull().sum())
```

```python
# ==========================================
# STEP 5: Check and Remove Duplicate Rows
# ==========================================

print("Duplicate Rows Before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate Rows After:", df.duplicated().sum())
print("Current Dataset Shape:", df.shape)
```

```python
# ==========================================
# STEP 6: Convert Transaction_Time
# ==========================================

df["Transaction_Time"] = pd.to_datetime(df["Transaction_Time"])

print("✅ Transaction_Time converted successfully!")
print(df["Transaction_Time"].head())
```

```python
# ==========================================
# STEP 7: Feature Engineering
# ==========================================

df["Transaction_Hour"] = df["Transaction_Time"].dt.hour
df["Transaction_Day"] = df["Transaction_Time"].dt.day
df["Transaction_Month"] = df["Transaction_Time"].dt.month

print("✅ New Features Created!")
df[["Transaction_Hour","Transaction_Day","Transaction_Month"]].head()
```

```python
# ==========================================
# STEP 8: Label Encoding
# ==========================================

label_encoder = LabelEncoder()

categorical_columns = [
    "Transaction_ID",
    "Customer_ID",
    "Gender",
    "Product_Category",
    "Payment_Method",
    "Device",
    "Browser",
    "Country",
    "IP_Address",
    "Shipping_Address_Match"
]

for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])

print("✅ Label Encoding Completed!")
```

```python
# ==========================================
# STEP 9: Drop Original Transaction_Time
# ==========================================

df.drop("Transaction_Time", axis=1, inplace=True)

print("Remaining Columns:")
print(df.columns)
```

```python
# ==========================================
# STEP 10: Save Processed Dataset
# ==========================================

df.to_excel("../Data/processed_fraud_dataset.xlsx", index=False)

print("✅ Processed dataset saved successfully!")
print("Shape:", df.shape)
```
